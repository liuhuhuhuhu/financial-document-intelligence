## 01. Document Parsing & Text Extraction

**Project:** Financial Document Intelligence System  
**Document:** Apple Inc. FY2025 Form 10-K  
**Tools:** Python, PyMuPDF, Google Colab

### Overview

This notebook implements the first stage of an end-to-end
financial document intelligence pipeline.

The objective is to extract and organize financial report
text for downstream semantic search and RAG applications.

### Objectives

1. Load publicly available SEC financial reports.
2. Extract text using PyMuPDF.
3. Preserve page-level metadata for source attribution.
4. Prepare structured documents for text chunking.

### Pipeline

PDF Document → Text Extraction → Page Metadata
→ Text Chunking → Embeddings → Vector Search → RAG

**Current stage:** Document parsing and text extraction.

```python
from google.colab import files

uploaded = files.upload()
```

```python
import pymupdf
from pathlib import Path

pdf_name = next(
    name for name in uploaded
    if name.lower().endswith(".pdf")
)

pages = []

with pymupdf.open(pdf_name) as doc:
    total_pages = len(doc)

    for i, page in enumerate(doc):
        text = page.get_text("text")

        if text.strip():
            pages.append({
                "page": i + 1,
                "text": text.strip()
            })

print("Total PDF pages:", total_pages)
print("Pages with extracted text:", len(pages))

if pages:
    print("\nPreview:\n")
    print(pages[0]["text"][:1500])
else:
    print("No extractable text found. OCR may be needed.")
```
